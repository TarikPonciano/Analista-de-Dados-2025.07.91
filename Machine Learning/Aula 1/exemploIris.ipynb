# DataSet Iris - KNN


```python
# pip install scikit-learn

import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["tipo"] = iris.target

dados_preditivos = df[iris.feature_names]
alvo = df["tipo"]

dados_preditivos_train, dados_preditivos_teste, alvo_train, alvo_teste =  train_test_split(dados_preditivos, alvo, test_size=0.3, random_state=42)

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(dados_preditivos_train, alvo_train)

alvo_previsto = knn.predict(dados_preditivos_teste)

acuracia = accuracy_score(alvo_teste, alvo_previsto)

display(f'Precisão do KNN k=3: {acuracia*100:.2f}%')

display(df)
```
