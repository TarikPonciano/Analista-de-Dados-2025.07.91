# Verificar se a pessoa vai a praia - KNN


1. Transformar os dados em dataframe
    - Nomear as colunas ("Temperatura", "Umidade", "Alvo")
2. Separar os dados em dados_preditivos e alvos
3. Produzir os dados de treinamento e os dados de teste
4. Criar o modelo knn
5. Treinar o modelo knn com os dados de treinamento
6. Verificar a precisão do modelo

```python
# pip install scikit-learn

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
dados = [
    [18, 90, 0], [20,85,0], [22,80,0], [25,70,1],
    [27, 65, 1], [30,50,1], [15,95,0], [28,55,1],
    [23, 75, 1], [19,88,0], [26,60,1], [21,82,0]
]

df = pd.DataFrame(dados, columns=["Temperatura", "Umidade", "Alvo"])

dados_preditivos = df[["Temperatura", "Umidade"]]

alvos = df["Alvo"]

dados_preditivos_treino, dados_preditivos_teste, alvos_treino, alvos_teste = train_test_split(dados_preditivos, alvos, test_size=0.3, random_state=42)

knn = KNeighborsClassifier(n_neighbors=3)

knn.fit(dados_preditivos_treino, alvos_treino)

alvos_previsao = knn.predict(dados_preditivos_teste)

precisao = accuracy_score(alvos_teste, alvos_previsao)

display(f"Precisão do Algoritmo = {precisao*100:.2f}%")

# Produzir gráfico resultado do teste
df_previsto = dados_preditivos_teste
df_previsto["Alvo"] = alvos_previsao
df_previsto["Alvo Verdadeiro"] = alvos_teste

display(df_previsto)
```
